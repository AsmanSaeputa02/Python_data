```python
# 1. สร้างตัว Decorator (คนที่จะเอาไปห่อคนอื่น)
def my_decorator(func):
    def wrapper():
        print("--- ก่อนเริ่มทำงาน ---")
        func() # เรียกใช้ฟังก์ชันเดิม
        print("--- ทำงานเสร็จแล้ว ---")
    return wrapper

# 2. นำไปใช้งานด้วย @
@my_decorator
def say_hello():
    print("สวัสดีจ้า!")

# 3. ลองเรียกใช้
say_hello()
```

```python
def login_required(func):
    def wrapper(user):
        if user == "admin":
            return func(user)
        else:
            return "❌ เข้าไม่ได้! คุณไม่ใช่แอดมิน"
    return wrapper

@login_required
def delete_database(user):
    return "🗑️ ลบฐานข้อมูลเรียบร้อยแล้ว"

# ลองใช้งาน
print(delete_database("guest")) # ผลลัพธ์: เข้าไม่ได้!
print(delete_database("admin")) # ผลลัพธ์: ลบฐานข้อมูลเรียบร้อยแล้ว
```

```python
# นี่คือ Decorator (กล่องห่อ)
def wash_hands_decorator(func):
    def wrapper():
        print("🧼 ล้างมือก่อน...") # (1) สิ่งที่เพิ่มเข้าไปก่อน
        func()                  # (2) เรียกฟังก์ชันเดิม (ทำกะเพรา)
        print("🍽️ พร้อมเสิร์ฟ!")    # (3) สิ่งที่เพิ่มเข้าไปหลัง
    return wrapper

@wash_hands_decorator # แปะป้ายบอกว่า "ห่อฟังก์ชันนี้ด้วยนะ"
def cook():
    print("♨️ กำลังผัดกะเพรา")

cook()
```

```python
import functools

# 1. ตัว Decorator (แม่พิมพ์)
def star_decorator(func):
    # functools.wraps ช่วยให้ฟังก์ชันเดิมไม่ลืม "ชื่อ" ของตัวเอง
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        # --- ก่อนรันฟังก์ชันหลัก ---
        print("⭐️" * 10) 
        
        # --- รันฟังก์ชันหลัก (รับค่าทุกอย่างที่ส่งมา) ---
        result = func(*args, **kwargs)
        
        # --- หลังรันฟังก์ชันหลัก ---
        print("⭐️" * 10)
        
        return result # ส่งค่าที่ได้จากฟังก์ชันหลักกลับออกไป
    return wrapper

# 2. ลองใช้งานกับฟังก์ชันที่ "ไม่มี" ตัวแปร
@star_decorator
def say_hi():
    print("สวัสดีจ้า!")

# 3. ลองใช้งานกับฟังก์ชันที่ "มี" ตัวแปร
@star_decorator
def greet(name, message="ยินดีที่ได้รู้จัก"):
    print(f"คุณ {name} บอกว่า: {message}")

# --- ทดสอบการรัน ---
say_hi()
print("\n") # เว้นวรรคเฉยๆ
greet("สมชาย", message="กินข้าวหรือยัง?")
```

```python
## decorator with argument

def repeat(n):
    def decorator(func):
        def wrapper( *args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
          
```

```python
@repeat(3)
def say_hey():
    print("สวัสดีจ้า!")
```

```python
say_hey()
```
