```python
import sqlite3
```

```python
## connect to an SQLite database

connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## create a table
cursor.execute('''
Create Table If not Exists employees(
    id Integer Primary key,
    name Text Not Null,
    age Integer,
    department text
               )
''')

# commit  the changer
connection.commit()
```

```python
cursor.execute(
    '''
Select * from employees
'''
)
```

```python
# insert the data in sqlite table

cursor.execute('''
Insert Into employees(name,age,department)
               values('asman',23,'data sci')
'''
)

cursor.execute('''
Insert Into employees(name,age,department)
               values('ryo',23,'gamer')
'''
)


cursor.execute('''
Insert Into employees(name,age,department)
               values('josh',84,'teacher')
'''
)

#commit the change
connection.commit()


```

```python
## query the data from the table

cursor.execute('select * from employees')
rows=cursor.fetchall()

### print the quried data

for row in rows:
    print(row)
```

```python
## update data in the table 

cursor.execute(
    '''
UPDATE employees
set age=24 
where name='asman'
'''
)
connection.commit
```

```python
## delete the data from the table
cursor.execute(
    '''
Delete from employees
where id = 5
'''
)
connection.commit()
```

```python
import sqlite3

connection = sqlite3.connect('sale_data.db')
cursor = connection.cursor()

# --- ขั้นตอนพิเศษ: ลบตารางเดิมทิ้งก่อนเพื่อ Reset โครงสร้าง ---
cursor.execute('DROP TABLE IF EXISTS sales') 

# --- สร้างตารางใหม่ที่มีคอลัมน์ครบถ้วน ---
cursor.execute('''
CREATE TABLE sales(
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')

sales_data = [
    ('2025-01-01', 'Product1', 100, 'North'),
    ('2025-01-02', 'Product1', 120, 'North'),
    ('2025-01-03', 'Product1', 110, 'North'),
    ('2025-01-04', 'Product1', 130, 'North'),
    ('2025-01-05', 'Product1', 170, 'North'),
    ('2025-01-06', 'Product1', 180, 'North')
]

# ใส่ข้อมูล
cursor.executemany('''
    INSERT INTO sales(date, product, sales, region)
    VALUES(?, ?, ?, ?)
''', sales_data)

connection.commit()
print("สร้างตารางและใส่ข้อมูลใหม่เรียบร้อยแล้ว!")

connection.close()
```
