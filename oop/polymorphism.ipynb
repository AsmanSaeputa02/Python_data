```python
class BankAccount:
    def display_status(self):
        print("Checking balance...")

class Firewall:
    def display_status(self):
        print("Checking security rules...")

# ฟังก์ชันกลางที่รับ "อะไรก็ได้" มาแสดงสถานะ
def show_report(obj):
    obj.display_status() # ไม่สนว่าเป็น Class ไหน ขอแค่มีฟังก์ชันนี้ก็พอ

show_report(BankAccount()) # ผลลัพธ์: Checking balance...
show_report(Firewall())    # ผลลัพธ์: Checking security rules...
```

```python
class Animal:
    def speak(self):
        pass # คลาสแม่มีโครงสร้างไว้ให้

class Dog(Animal):
    def speak(self):
        return "Woof! Woof!" # ลูกเขียนทับให้เป็นเสียงตัวเอง

class Cat(Animal):
    def speak(self):
        return "Meow! Meow!" # ลูกอีกตัวก็เขียนทับในแบบของตัวเอง

# การใช้งาน Polymorphism
animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak()) # ใช้คำสั่งเดียวกัน (speak) แต่ผลลัพธ์ต่างกัน
```

```python
#polimorphism  with Functions and Methods
## base class


class Shape:
    def area(self):
        return "The area of the figure "
    
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height # ทับฟังก์ชัน area ในคลาส
    
    ## derived class 2
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2 # ทับฟังก์ชัน area ในคลาส

# การใช้งาน Polymorphism กับฟังก์ชัน
def calculate_area(shape):
    print(shape.area()) # เรียกใช้ฟังก์ชัน area ของแต่ละคลาส
    
rect = Rectangle(5, 3)
circle = Circle(4)
calculate_area(rect)   # ผลลัพธ์: 15
calculate_area(circle) # ผลลัพธ์: 50.24
```
