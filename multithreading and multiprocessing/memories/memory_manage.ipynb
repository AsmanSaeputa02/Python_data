```python
import gc 


class Myobject:
    def __init__(self,name):
        self.name =name
        print(f"Object {self.name} created")
    
    def __del__(self):
        print(f"Object {self.name} deleted")


# Creaate circular refreance

obj1 = Myobject("obj1")
obj2 = Myobject("obj2")

obj1.ref = obj2
obj2.ref = obj1


del obj1
del obj2


### manually trigger yhe garbage collection

gc.collect()
```
