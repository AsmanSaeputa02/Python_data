```python
# create set
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
empty_set = set()
print(type(empty_set))
```

```python
my_set = set([1,2,3,6,4,5,6,4])
print(my_set)
```

```python
#basic set operations
# adding  removing elements
my_set.add(7)
print(my_set)
my_set.remove(2)
print(my_set)
```

```python
my_set.discard(10)  # does not raise an error if the element is not present
print(my_set)
```

```python
## pop method

removed_element = my_set.pop()
print(f"Removed element: {removed_element}")
print(my_set)
```

```python
## clear elements from set
my_set.clear()
print(my_set)
```

```python
# set membership
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # True
print(6 in my_set)  # False
```

```python
## math operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# union
union_set = set_a.union(set_b)
print(f"Union: {union_set}")
# iNtersection
intersection_set = set_a.intersection(set_b)
print(f"Intersection: {intersection_set}")

# difference
difference_set = set_a.difference(set_b)
print(f"Difference (A - B): {difference_set}")

# symmetric difference
sym_diff_set = set_a.symmetric_difference(set_b)
print(f"Symmetric Difference: {sym_diff_set}")

```

```python
# set methods
set_a = {1, 2, 3, 4,5,6}
set_b = {3, 4, 5, 6}

# is subset
print(set_a.issubset(set_b))  
# is superset
print(set_a.issuperset(set_b))
```

```python
## counting  unique  words in a text
text = "in this tuutorial we are dicussing aout sets in python sets are unordered collections of unique elements"
words = text.split()

#convert list of words to a set to get unique words

unique_words = set(words)
print(f"Unique words: {unique_words}")
print(f"Number of unique words: {len(unique_words)}")
```
