```python
# inheritance
# parent class

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    def drive(self):
        print(f"The person is driving the  {self.enginetype} car")
```

```python
car1 = Car(5,6,"petrol")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,autopilot):
        super().__init__(windows,doors,enginetype)
        self.autupilot = autopilot
    def autopilot_drive(self):
        print(f"The person is driving the  car with autopilot {self.autupilot}")
```

```python
tesla1 = Tesla(4,4,"electric",True)
tesla1.autopilot_drive()
```

```python
tesla1.drive()
```
