```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __truediv__(self, scalar):
        return Vector(self.x / scalar, self.y / scalar)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    def __eq__(self, value):
        return self.x == value.x and self.y == value.y
    
# create objects of the Vector class
v1 = Vector(2, 3)
v2 = Vector(4, 5)
print(v1+v2)  # Output: Vector(2, 3)
```

```python
class Weapon:
    def __init__(self, name, damage):
        self.name = name
        self.damage = damage

    # Overloading เครื่องหมาย +
    def __add__(self, other):
        # self คือ ดาบเล่มแรก, other คือ ดาบเล่มที่สอง
        new_damage = self.damage + other.damage
        return Weapon(f"Combined {self.name}", new_damage)

    # Overloading เครื่องหมาย > (เปรียบเทียบความแรง)
    def __gt__(self, other):
        return self.damage > other.damage

    def __str__(self):
        return f"{self.name} (Damage: {self.damage})"

# --- การใช้งาน ---
sword1 = Weapon("ดาบไม้", 10)
sword2 = Weapon("ดาบเหล็ก", 50)

# ใช้ + ระหว่าง Object ได้เลย!
mega_sword = sword1 + sword2 
print(mega_sword) # ผลลัพธ์: Combined ดาบไม้ (Damage: 60)

# ใช้ > เปรียบเทียบได้เลย!
if sword2 > sword1:
    print(f"{sword2.name} แรงกว่า {sword1.name}")
```
