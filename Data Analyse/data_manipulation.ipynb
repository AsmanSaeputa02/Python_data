```python
# Handling Missing Values
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_fillna = df.fillna(0)
```

```python
## filling missin values with the mean of the coilumn
# df['Sales_fillNA']=df['category'].fillna(df['category'].mean())
# เติมด้วยค่าที่ฮิตที่สุดในหมวดหมู่
most_frequent = df['category'].mode()[0] 
df['category_filled'] = df['category'].fillna(most_frequent)
df
```

```python
df['unit_price_filled']=df['unit_price'].fillna(df['unit_price'].mean())
```

```python
df.head(5)
```

```python
df.dtypes
```

```python
## Reaming column
df.rename(columns={'date':'date_date'})
df.head(5)
```

```python
## change datatype
# df['Price']=df['unit_price'].astype(float)
df['Price']=df['unit_price'].fillna(df['unit_price'].mean()).astype(int)
df.head(1)
```

```python
df['new_price']=df['unit_price'].apply(lambda x:x*2)
df.head(5)
```

```python
## Das Agregating and Grouping
group_mean = df.groupby('product_name')['Price'].mean()
group_mean
```

```python
df.groupby(['product_name','category'])['quantity'].sum()
```

```python
# arggregate multi function 
# สรุปข้อมูลคอลัมน์ total_amount โดยหาทั้ง ผลรวม, ค่าเฉลี่ย, และค่ามากสุด
df.groupby('category')['total_amount'].agg(['sum', 'mean', 'max'])
```

```python
## Merging and join dataframe


df1= pd.DataFrame({'key':['A','B','c'], 'Value1': [1,2,3]})
df2= pd.DataFrame({'key':['A','B','d'], 'Value1': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge Dataframe on the  ' key columns  '


pd.merge(df1,df2,on="key",how="inner")
```

```python
pd.merge(df1,df2,on="key",how="outer")
```

```python
pd.merge(df1,df2,on="key",how="left")
```

```python
pd.merge(df1,df2,on="key",how="right")
```
