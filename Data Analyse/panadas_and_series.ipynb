```python
#series คือโครงสร้างข้อมูลแบบหนึ่งใน Pandas ที่ใช้เก็บข้อมูลในรูปแบบของลำดับ (1D) โดยมีการจัดเรียงข้อมูลเป็นลำดับและมีป้ายกำกับ (index) ที่สามารถใช้ในการเข้าถึงข้อมูลได้อย่างง่ายดาย


import pandas as pd 
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
# create a series from a dictionary

data_dict = {'a': 1, 'b': 2, 'c': 3}
series_from_dict = pd.Series(data_dict)
print( series_from_dict)
```

```python
data = [10, 20, 30, 40, 50]
index = ['a', 'b', 'c', 'd', 'e']
pd.Series(data, index=index)
```

```python
# dataframe
#create dataframe from  dictionary

data = {
    'Name':['asman','john','jack'],
    'Age':[25,30,45],
    'City':['Bankok','New York' , 'Florida']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```
